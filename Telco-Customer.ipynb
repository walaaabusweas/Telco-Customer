3) Fix inconsistent categories

Now I will  create the new Features

2) (New Feature)  is_Fiber_optic_subscriber 

```python
df["is_Fiber_optic_subscriber"] = df["InternetService"].apply(lambda x: "Yes" if x == "Fiber optic" else "No")
```

3) (New Feature) is_Electronic_check_Payment

```python
df["is_Electronic_check_Payment"] = df["PaymentMethod"].apply(lambda x: "Yes" if x == "Electronic check" else "No")
```

4) (New Feature) is_Active_Digital_User 

```python
df["is_Active_Digital_User"] = df.apply(
    lambda row: "Yes" if row["PaperlessBilling"] == "Yes" and row["SeniorCitizen"] == "No" else "No", 
    axis=1
)
```

5) (New Feature) is_new_customer

```python
df["is_new_customer"] = df["tenure"].apply(lambda x: "Yes" if x <= 10 else "No")
```

6) (New Feature) avg_spend_per_month

```python
# use this logic TotalCharges / (tenure + 1) in apply
df["avg_spend_per_month"] = df.apply(lambda row: row["TotalCharges"] / (row["tenure"] + 1), axis=1)
```

7) (New Feature) youth_loyalty_indicator 

```python
# Yes if user is not senior (High Loyalty Potential) , and No otherwise
df["youth_loyalty_indicator"] = df["SeniorCitizen"].apply(lambda x: "No" if x == "Yes" else "Yes")
```

8) (New Feature) is_streaming-Fiber-subscriber.

```python
# use yes if it is (Fiber optic + StreamingTV/Movies), and No for everything else
df["is_streaming-Fiber-subscriber"] = df.apply(lambda row: "Yes" if (row["InternetService"] == "Fiber optic" and (row["StreamingTV"] == "Yes" or row["StreamingMovies"] == "Yes")) else "No", axis=1)
```
