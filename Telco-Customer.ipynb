Hypothesis 2: The Fiber Optic Churn Crisis

The hypothesis: Although fiber optic service offers the best quality internet, it has the highest churn rates compared to DSL services or customers without internet.

Supporting Data: * When analyzing the "InternetService" , we find that a very large percentage of departing customers (Churn = Yes) belong to this specific category("Fiber Optic).

Fiber_Churn = df[["Churn","InternetService"]]

count_Fiber_Churn.value_counts()

Explanation:
There's a strange contradiction here; Viber is faster, yet it drives customers away! The logical explanation is that these customers are paying high prices for the service, and if the "value for money" isn't clear to them, or if they encounter even a minor technical issue, they leave immediately because the high bill makes them unwilling to tolerate any mistakes. Furthermore, the company's competitors often target Viber users with very attractive "switch" offers.

Business Insight:
The company should immediately conduct an exit survey with these customers to determine whether the reason for leaving was price or underlying technical issues. Price lock packages should also be offered to fiber customers to ensure they don't switch to competitors to save a few dollars.
